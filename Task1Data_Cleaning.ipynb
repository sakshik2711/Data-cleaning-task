```python
import pandas as pd
import numpy as np
```

```python
# Load dataset
df = pd.read_csv("Sales-Export_2019-2020.csv")

# Show first 5 rows
df.head()
```

```python
# Dataset info
df.info()

# Check shape
df.shape

# Check column names
df.columns
```

```python
# Check null values
df.isnull().sum()
```

```python
# Example: Fill numeric columns with mean
df.fillna(df.mean(numeric_only=True), inplace=True)

# For text columns, fill with 'Unknown'
df.fillna("Unknown", inplace=True)
```

```python
df.isnull().sum()
```

```python
# Check duplicates
df.duplicated().sum()

# Remove duplicates
df.drop_duplicates(inplace=True)
```

```python
# Make column names lowercase and remove spaces
df.columns = df.columns.str.lower().str.replace(" ", "_")

df.columns
```

```python
# Clean column names
df.columns = df.columns.str.strip().str.lower()
df.columns = df.columns.str.replace(" ", "_")
df.columns = df.columns.str.replace("^_|_$", "", regex=True)

print(df.columns)
```

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')
```

```python
df['order_value_eur'] = pd.to_numeric(df['order_value_eur'], errors='coerce')
df['cost'] = pd.to_numeric(df['cost'], errors='coerce')
```

```python
# Example: make country names uppercase
df['country'] = df['country'].str.upper()

# Remove extra spaces
df['country'] = df['country'].str.strip()
```

```python
df.drop_duplicates(inplace=True)
```
